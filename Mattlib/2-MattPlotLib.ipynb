```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data1 = np.linspace(0,10,20)
```

```python
data2 = data1**2
```

```python
plt.subplots()
```

```python
my_fig, my_axes = plt.subplots()
```

```python
my_fig, my_axes = plt.subplots()
my_axes.plot(data1,data2,color="#1E0703",alpha=0.5)
my_axes.plot(data2,data1)
plt.show()
```

```python
my_fig, my_axes = plt.subplots()
my_axes.plot(data1,data1+2,color="blue",linewidth=1.0)
my_axes.plot(data1,data1+3,color="yellow",linewidth=5.0)
my_axes.plot(data1,data1+4,color="yellow",linestyle = "-.")
my_axes.plot(data1,data1+5,color="yellow",linestyle = "--")
my_axes.plot(data1,data1+6,color="yellow",linestyle = "--",marker ="o" ,markersize=8,markerfacecolor="red")
plt.show()
```

```python
#scatter
plt.scatter(data1,data2)
plt.show()
```
