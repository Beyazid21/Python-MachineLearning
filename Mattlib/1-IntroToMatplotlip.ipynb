```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
age_list = [10,20,30,30,30,50,50,60,70,75]
```

```python
weight_list = [20,60,80,85,86,87,70,90,95,90]
```

```python
plt.plot(age_list,weight_list,"r") #y-yellow ,g-green r-red
plt.xlabel("Age")
plt.ylabel("Weight")
plt.savefig("test.png")
```

```python
np_age_list = np.array(age_list)
```

```python
np_weight_list = np.array(weight_list)
```

```python
plt.plot(np_age_list,np_weight_list,"g")
plt.xlabel("Age")
plt.ylabel("Weight")
```

```python
numpy_arr1 = np.linspace(0,10,20) #0dan 10 qədər 20 ədəd yarat arasındaki məsafələr eyni olsun
```

```python
numpy_arr1
```

```python
numpy_arr2 = numpy_arr1 **3
```

```python
numpy_arr2
```

```python
plt.plot(numpy_arr1,numpy_arr2,"b--")
plt.show()
```

```python
#Iki qrafiki yan yana qoymaq
```

```python
plt.subplot(1,2,1)
plt.plot(numpy_arr1 , numpy_arr2,"r*-")
plt.subplot(1,2,2)
plt.plot(numpy_arr2 , numpy_arr1,"g--")
plt.show()
```

```python
#Fiqur gostermek
```

```python
my_figure = plt.figure()
figure_axes = my_figure.add_axes([0.1,0.1,0.3,0.3]) #birinci ve ikinci reqem soldan ve sagdan bosluqu 3-4 cu ise qrafikin ölçüsün teyin edir
figure_axes.plot(numpy_arr1,numpy_arr2,"g")
figure_axes.set_xlabel("X Axsis")
figure_axes.set_ylabel("Y Axsis")
figure_axes.set_title("Graf title")
plt.show()
```

```python
new_fig = plt.figure(dpi = 100)
new_axes = new_fig.add_axes([0.1,0.1,0.9,0.9])
new_axes.plot(numpy_arr1,numpy_arr1**2,label="numpy array**2")
new_axes.plot(numpy_arr1,numpy_arr1**3,label="numpy array**3")
new_axes.legend()
plt.show()
```
