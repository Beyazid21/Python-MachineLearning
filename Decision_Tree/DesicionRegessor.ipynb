```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('life expectancy.csv')
```

```python
df.head()
```

```python

df.info()
df.isnull().sum()
```

```python
#As seen, there are no missing values in the dataset, and all data types are correctly formatted.
```

```python
df.Year.value_counts()
```

```python
#Let's separate our target and independent variables.
X = df.drop(["Country Name","Country Code","Year","Life_Expectancy","index"], axis = 1)
y = df["Life_Expectancy"]
            
```

```python
X.head()
```

```python
#Splitting the dataset into training and testing sets.
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3 ,random_state=15)
```

```python
from sklearn.tree  import DecisionTreeRegressor
```

```python
#Grid Search
#Defining the parameter dictionary for Hyperparameter Tuning (GridSearchCV)
param = {
    "criterion":["squared_error", "friedman_mse", "absolute_error", "poisson"],
    "splitter":["best","random"],
    "max_depth":[1,2,3,5,10,15,20,30,None],
    "max_features":["sqrt","log2",None],
    "ccp_alpha" :[0.01,0.0,0.1,0.001,1,10]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(estimator=DecisionTreeRegressor(), param_grid=param, cv = 5 , scoring="r2")
```

```python
import warnings
warnings.filterwarnings("ignore")
grid.fit(X_train,y_train) #Initializing and starting the model training process.
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test) 
```

```python
from sklearn.metrics import  r2_score
```

```python
print("r2_score:",r2_score(y_test,y_pred))
```

```python
# Predicting life expectancy for Azerbaijan using 2024-2025 estimated data
# Feature order: [PM25, Population_Density, Stability, Education, Health_Expenditure, Urban_Rate]
azerbaijan_data = pd.DataFrame([[12, 126, -0.73, 42, 304, 58.5]], columns=X.columns)

# Real-world average life expectancy for Azerbaijan: ~73.5 - 74.2 years
y_pred_az = grid.predict(azerbaijan_data)

print(f"Predicted Life Expectancy for Azerbaijan: {y_pred_az[0]:.2f} years")
print(f"Real-world Reference: ~74.0 years")
```

```python
from sklearn.ensemble import RandomForestRegressor

# Define a broader hyperparameter grid for the Random Forest model
# n_estimators: Number of trees in the forest
# max_depth: Maximum depth of each tree
# min_samples_split: Minimum number of samples required to split an internal node
rf_params = {
    'n_estimators': [100, 200, 300], 
    'max_depth': [5,10, 20, None],
    'min_samples_split': [2, 5, 10,20,30]
}

# Initialize GridSearchCV with RandomForestRegressor
# We use 5-fold cross-validation to ensure model stability across different data splits
rf_grid = GridSearchCV(RandomForestRegressor(random_state=42), rf_params, cv=5, scoring='r2', n_jobs=-1)

# Fit the model to the training data to find the optimal forest configuration
rf_grid.fit(X_train, y_train)

# Output the best R2 score achieved during the search
print(f"Best R2 Score after Random Forest Tuning: {rf_grid.best_score_:.4f}")
```

```python
y_pred_new = rf_grid.predict(X_test)
print("r2_score:",r2_score(y_test,y_pred_new))
```
