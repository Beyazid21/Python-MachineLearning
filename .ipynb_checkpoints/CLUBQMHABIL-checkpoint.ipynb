```python
import pandas as pd
import numpy as np
import requests
import firebase_admin
from firebase_admin import credentials, firestore

# --- 1. AYARLAR ---
XACMAZ_LAT = 41.46
XACMAZ_LON = 48.81
CLUB_ID = '05yxI5swDM02L2zhMV5u'
JSON_KEY_PATH = '/home/bayazit/Downloads/globalclubmanegment-firebase-adminsdk-fbsvc-31f6d35b44.json'

if not firebase_admin._apps:
    cred = credentials.Certificate(JSON_KEY_PATH)
    firebase_admin.initialize_app(cred)
db = firestore.client()

# --- 2. DATA FETCH & CLEANING ---
docs = db.collection('club_daily_session_summaries') \
    .where('clubId', '==', CLUB_ID) \
    .where('totalRevenue', '>=', 32) \
    .stream()
rows = [doc.to_dict() for doc in docs]
df = pd.DataFrame(rows)

drop_cols = ['cancelledCount','clubId','completedCount','timeRevenue',
             'productRevenue','productCost','updatedAt','billableMinutes','totalMinutes', 'id']
df = df.drop(columns=[c for c in drop_cols if c in df.columns])

df['dayKey'] = pd.to_datetime(df['dayKey'])
df = df.sort_values('dayKey').reset_index(drop=True)

# --- 3. TƏKMİLLƏŞDİRİLMİŞ TƏTİL METODU ---
def check_all_holidays(date):
    m = date.month
    d = date.day
    y = date.year
    dw = date.dayofweek # 5=Şənbə, 6=Bazar
    
    is_public = 0
    is_school = 0
    is_uni = 0
    is_exam = 0
    
    # 3a. Rəsmi Bayramlar
    fixed_holidays = [
        (1, 1), (1, 2), (1, 20), (3, 8), 
        (3, 20), (3, 21), (3, 22), (3, 23), (3, 24), 
        (5, 9), (5, 28), (6, 15), (6, 26), (10, 18), 
        (11, 8), (11, 9), (11, 12), (11, 17), (12, 31)
    ]
    if (m, d) in fixed_holidays:
        is_public = 1
        
    if y == 2026:
        if m == 3 and d in [20, 21]: is_public = 1 # Ramazan
        if m == 5 and d in [27, 28]: is_public = 1 # Qurban

    # 3b. Həftə sonu faktoru (Şənbə=5, Bazar=6)
    is_weekend = 1 if dw in [5, 6] else 0

    # 3c. MƏKTƏB TƏTİLİ (Bayram + Weekend + Akademik)
    if is_public == 1 or is_weekend == 1:
        is_school = 1
    elif m == 11 and 10 <= d <= 14: is_school = 1
    elif (m == 12 and d >= 22) or (m == 1 and d <= 5): is_school = 1
    elif m == 3 and 23 <= d <= 27: is_school = 1
    elif m == 5 and 4 <= d <= 8: is_school = 1
    elif (m == 6 and d >= 15) or (m in [7, 8]) or (m == 9 and d <= 14): is_school = 1

    # 3d. UNİVERSİTET TƏTİLİ (Bayram + Weekend + Akademik)
    if is_public == 1 or is_weekend == 1:
        is_uni = 1
    elif (m == 1 and d >= 29) or (m == 2 and d <= 15): is_uni = 1
    elif (m == 6 and d >= 26) or (m in [7, 8]) or (m == 9 and d <= 14): is_uni = 1
    
    # 3e. İmtahanlar
    if m == 1 and 5 <= d <= 28: is_exam = 1
    elif m == 6 and 1 <= d <= 25: is_exam = 1

    return pd.Series([is_public, is_school, is_uni, is_exam])

# Metodu tətbiq edirik
df[['is_public_holiday', 'is_school_holiday', 'is_university_holiday', 'is_uni_exam_period']] = df['dayKey'].apply(check_all_holidays)

# --- 4. HAVA DURUMU ---
start_date = df['dayKey'].min().strftime('%Y-%m-%d')
end_date = df['dayKey'].max().strftime('%Y-%m-%d')
weather_url = f"https://archive-api.open-meteo.com/v1/archive?latitude={XACMAZ_LAT}&longitude={XACMAZ_LON}&start_date={start_date}&end_date={end_date}&daily=temperature_2m_mean,precipitation_sum,shortwave_radiation_sum&timezone=auto"

try:
    w_res = requests.get(weather_url).json()
    w_df = pd.DataFrame({
        'dayKey': pd.to_datetime(w_res['daily']['time']),
        'avg_temp': w_res['daily']['temperature_2m_mean'],
        'rain_intensity': w_res['daily']['precipitation_sum'],
        'daylight_strength': w_res['daily']['shortwave_radiation_sum']
    })
    df = pd.merge(df, w_df, on='dayKey', how='left')
except:
    print("Hava məlumatı alınmadı.")

# --- 5. FEATURE ENGINEERING ---
df['day_of_week'] = df['dayKey'].dt.dayofweek
df['month'] = df['dayKey'].dt.month
df['is_weekend'] = df['day_of_week'].isin([5, 6]).astype(int)
df['is_holiday_eve'] = df['is_public_holiday'].shift(-1, fill_value=0)
df['is_after_holiday'] = df['is_public_holiday'].shift(1, fill_value=0)

# Lag və Rolling (Sıfırları doldurmaq üçün öncə median hesablayırıq)
df['prev_day_revenue'] = df['totalRevenue'].shift(1).fillna(df['totalRevenue'].median())
df['prev_week_revenue'] = df['totalRevenue'].shift(7).fillna(df['totalRevenue'].median())
df['rolling_7d_avg'] = df['totalRevenue'].shift(1).rolling(window=7).mean().fillna(df['totalRevenue'].median())

# --- 6. EXPORT VƏ ANALİZ ---
df.to_csv('club_analysis_fixed.csv', index=False)

print("\nYeni Məntiq ilə Analiz:")
comparison = df.groupby('is_university_holiday')['totalRevenue'].agg(['mean', 'median', 'count'])
comparison.index = ['Dərs Günləri (Həftə içi)', 'Tətil/Həftəsonu']
print(comparison)

print("\nDatasetin ilk 10 sətri:")
print(df[['dayKey', 'totalRevenue', 'is_weekend', 'is_university_holiday']].head(10))
```

```python
import pandas as pd
import numpy as np

# 1. Mövcud datanı birbaşa yükləyirik (Sənin verdiyin data)
# (Bu hissəni öz df-inlə əvəz edə bilərsən)

# 2. Bazada olmayan (əvvəlki) tarixlər üçün gəlir təxminləri (Fevralın ilk həftəsi)
historical_revenues = {
    '2026-02-01': 82.50, # Bazar
    '2026-02-02': 61.20,
    '2026-02-03': 58.40,
    '2026-02-04': 55.10,
    '2026-02-05': 59.80,
    '2026-02-06': 64.30,
    '2026-02-07': 88.90, # Şənbə
    '2026-02-14': 78.40, # Aradakı boşluq (Şənbə)
    '2026-02-15': 81.20  # Aradakı boşluq (Bazar)
}

# 3. İLK SƏTİRLƏRİN DOLDURULMASI (Manual Imputation)

# Index 0: 2026-02-08 (Bazar)
df.at[0, 'prev_day_revenue'] = historical_revenues['2026-02-07']
df.at[0, 'prev_week_revenue'] = historical_revenues['2026-02-01']
df.at[0, 'rolling_7d_avg'] = 67.10 # (01-07 fevral ortalaması)

# Index 1-5 arası prev_week_revenue doldurma (7 gün əvvələ baxaraq)
for i in range(1, 6):
    target_date = df.at[i, 'dayKey'] - pd.Timedelta(days=7)
    df.at[i, 'prev_week_revenue'] = historical_revenues[target_date.strftime('%Y-%m-%d')]

# Index 6: 2026-02-16 (Bazar ertəsi)
# Cümə günü (13-ü) ilə bu gün arasında Şənbə/Bazar boşluğu var deyə əllə doldururuq:
df.at[6, 'prev_day_revenue'] = historical_revenues['2026-02-15']
df.at[6, 'prev_week_revenue'] = historical_revenues['2026-02-09'] if '2026-02-09' in historical_revenues else 63.78

# 4. ROLLING AVERAGE YENİDƏN HESABLAMA
# Sıfır qalan rolling_7d_avg sütunlarını ümumi median ilə doldururuq ki, model çaşmasın
median_revenue = df['totalRevenue'].median()
df['rolling_7d_avg'] = df['rolling_7d_avg'].replace(0, median_revenue)

print("İlk sətirlər uğurla dolduruldu.")
print(df[['dayKey', 'prev_day_revenue', 'prev_week_revenue', 'rolling_7d_avg']].head(7))
```

```python
df = df.drop('dayKey',axis = 1)
```

```python
X = df.drop('totalRevenue',axis = 1)
y = df['totalRevenue']
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 15)
```

```python
from xgboost import XGBRegressor
```

```python
model = XGBRegressor()
```

```python
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
# Proqnoz ver
preds = model.predict(X_test)

# Səhv payını hesabla (MAE)
mae = mean_absolute_error(y_test, preds)
r2 = r2_score(y_test, preds)

print(f"Ortalama Səhv (MAE): {mae:.2f} AZN")
print(f"R2 Score: {r2:.2f}")
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=50, max_depth=5, random_state=15)
model.fit(X_train, y_train)
```

```python

# Proqnoz ver
preds = model.predict(X_test)

# Səhv payını hesabla (MAE)
mae = mean_absolute_error(y_test, preds)
r2 = r2_score(y_test, preds)

print(f"Ortalama Səhv (MAE): {mae:.2f} AZN")
print(f"R2 Score: {r2:.2f}")
```
