```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('athlete_events.csv')
```

```python
plt.scatter("Height","Weight",data = data)
plt.xlabel("Height") 
plt.ylabel("Weight")
plt.title("Attlethe Height vs Weigh")
plt.show
```

```python
sns.set_style("whitegrid")
sns.scatterplot(x = "Height",y = "Weight",data=data)
plt.xlabel("Height of attlethe")
plt.ylabel("Weight of attlethe")
plt.title("Athlete Weight vs Height")
plt.show()
```

```python
sns.set_style("dark")
sns.scatterplot(x = "Height",y = "Weight",hue= "Sex",data=data)#Burada hue fətqli rənglərdə qruplaşdırır Sex in dəyərlərin
plt.xlabel("Height of attlethe")
plt.ylabel("Weight of attlethe")
plt.title("Athlete Weight vs Height")
plt.show()
```

```python
sns.set_style("dark")
sns.scatterplot(x = "Height",y = "Weight",hue= "Sex",style = "Medal",size="Age",data=data)
plt.xlabel("Height of attlethe")
plt.ylabel("Weight of attlethe")
plt.title("Athlete Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.lineplot(x="Height",y="Weight",hue = "Sex",data = data)
plt.xlabel("Height of attlethe")
plt.ylabel("Weight of attlethe")
plt.title("Athlete Weight vs Height")
plt.show()
```

```python
sns.set_style("dark")
sns.displot(x="Height",hue="Sex",data = data) #Displot paylanmani görmək üçün istifaə olunur
plt.ylabel("Frequency")
plt.title("Athletes Height Distrubutation")
plt.show()
```

```python
sns.set_style("dark")
sns.displot(x="Height",hue="Sex",data = data,kind = "kde") #Displot paylanmani görmək üçün istifaə olunur
plt.ylabel("Frequency")
plt.title("Athletes Height Distrubutation")
plt.show()
```

```python
sns.catplot(x = "Medal", y = "Height" , hue = "Sex" , col = "Season", data = data)
plt.show()
```
