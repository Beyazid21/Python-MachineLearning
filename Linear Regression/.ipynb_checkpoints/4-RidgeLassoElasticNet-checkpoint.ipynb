```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split (X,y,test_size = 0.25 , random_state = 15)
```

```python
# redundancy - > iki featurenin bir biri ilə çox corr olması
#multicolinearity,overfitting
```

```python
type(X_train.corr())
```

```python
print(X_train.corr().columns)
```

```python
print(X_train.corr().iloc[2,2])
```

```python
def correlection_for_dropping (df, threshold) :
    columns_to_drop = set()
    corr = df.corr()

    for i in range (len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i,j]) > threshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop
    
```

```python
columns_dropping = correlection_for_dropping(X_train,0.85)
```

```python
X_train.drop(columns_dropping, axis = 1,  inplace = True)
X_test.drop(columns_dropping, axis = 1,  inplace = True)
```

```python
X_train.head()
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
#scaler
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
plt.subplots(figsize = (15,5))
plt.subplot(1,2,1)
sns.boxplot(data = X_train)
plt.title("X_train")
plt.subplot(1,2,2)
sns.boxplot(data = X_train_scaled)
plt.title('X_train_scaled')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score
```

```python
linear = LinearRegression()
```

```python
linear.fit(X_train_scaled,y_train)
y_pred = linear.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test, y_pred)
print('Mean Absolute Error:',mae)
print('Mean squared Error:',mse)
print("R2 score :",score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
from sklearn.linear_model import Lasso
lasso = Lasso()
lasso.fit(X_train_scaled,y_train)
y_pred = lasso.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test, y_pred)
print('Mean Absolute Error:',mae)
print('Mean squared Error:',mse)
print("R2 score :",score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
from sklearn.linear_model import Ridge
ridge = Ridge()
ridge.fit(X_train_scaled,y_train)
y_pred = ridge.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test, y_pred)
print('Mean Absolute Error:',mae)
print('Mean squared Error:',mse)
print("R2 score :",score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
from sklearn.linear_model import ElasticNet
elastic = ElasticNet()
elastic.fit(X_train_scaled,y_train)
y_pred = elastic.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test, y_pred)
print('Mean Absolute Error:',mae)
print('Mean squared Error:',mse)
print("R2 score :",score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
#lasso cross validation
```

```python
from sklearn.linear_model import  LassoCV
lassocv = LassoCV(cv = 5)   
lassocv.fit(X_train_scaled,y_train)
y_pred = lassocv.predict(X_test_scaled)
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test, y_pred)
print('Mean Absolute Error:',mae)
print('Mean squared Error:',mse)
print("R2 score :",score)
plt.scatter(y_test,y_pred)
plt.show()
# cv=5 o deməkdir ki, X_train datası 5 hissəyə bölünür.
# LassoCV default olaraq 100 fərqli alpha dəyəri yaradır (n_alphas=100).
# Hər alpha dəyəri üçün model 5 dəfə fit olunur:
# 4 hissə train, 1 hissə validation kimi istifadə edilir.
# Bu proses 5 dəfə təkrarlanır və orta xəta hesablanır.
# Sonda bu 100 alpha arasından ən yaxşı nəticə verən alpha avtomatik seçilir.

```
