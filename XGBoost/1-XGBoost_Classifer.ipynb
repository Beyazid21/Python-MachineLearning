```python
from xgboost import XGBClassifier

# XGBoost Classifier Parametrləri Cheat Sheet
xgb_model = XGBClassifier(
    # =================================================================
    # 1. ÜMUMİ VƏ ÖYRƏNMƏ PARAMETRLƏRİ (GENERAL & LEARNING)
    # =================================================================

    # booster (Default: 'gbtree')
    # Hansı mühərrikin istifadə olunacağını seçir. 
    # 'gbtree' (ağac əsaslı) demək olar ki, həmişə ən yaxşı nəticəni verir.
    # Digər variantlar: 'gblinear' (xətti model), 'dart' (dropout ilə ağaclar)
    booster='gbtree',

    # objective (Default: 'binary:logistic')
    # Hədəf funksiyası (Biz nəyi həll edirik?).
    # - İkili təsnifat (Məs: Spam vs Normal): 'binary:logistic' (Ehtimal qaytarır)
    # - Çoxsınıflı (Məs: Qırmızı, Yaşıl, Mavi): 'multi:softprob' (Hər sinif üçün ehtimal qaytarır)
    objective='binary:logistic',

    # learning_rate / eta (Default: 0.3)
    # Öyrənmə sürəti. Hər yeni ağacın modelə qatqısını müəyyən edir.
    # Adətən 0.01 ilə 0.2 arası seçilir. Kiçik rəqəm overfitting-i azaldır, 
    # amma daha çox n_estimators (ağac sayı) tələb edir.
    learning_rate=0.1,

    # n_estimators (Default: 100)
    # Qurulacaq ağacların maksimum sayı (və ya boosting mərhələləri).
    # Çox böyük olması əzbərləməyə (overfitting) səbəb ola bilər. 
    # Mütləq "Early Stopping" ilə birlikdə istifadə edilməlidir.
    n_estimators=300,

    # =================================================================
    # 2. AĞACIN STRUKTURUNU TƏNZİMLƏYƏN PARAMETRLƏR (TREE CONFIG)
    # =================================================================

    # max_depth (Default: 6)
    # Bir ağacın maksimum dərinliyi. Modeli mürəkkəbləşdirir.
    # Overfitting varsa, bu rəqəmi azalt (məs: 3, 4 və ya 5).
    max_depth=5,

    # min_child_weight (Default: 1)
    # Bir yarpaqda (leaf node) olması tələb olunan minimum nümunə "ağırlığı".
    # Overfitting-in qarşısını almaq üçün ƏN VACİB parametrlərdən biridir. 
    # Modelin çox spesifik dataları əzbərləməsinin qarşısını almaq üçün artır (məs: 3, 5).
    min_child_weight=1,

    # gamma / min_split_loss (Default: 0)
    # Bir düyünün bölünməsi üçün tələb olunan minimum "itki azalması" (loss reduction).
    # Rəqəm artdıqca alqoritm daha konservativ olur (bölünməkdən çəkinir). 
    # Tuning zamanı 0.1, 0.2 kimi rəqəmlər yoxlanıla bilər.
    gamma=0.1,

    # =================================================================
    # 3. TƏSADÜFİLİK VƏ NÜMUNƏLƏMƏ (RANDOMNESS & SAMPLING)
    # =================================================================

    # subsample (Default: 1.0)
    # Hər ağac qurularkən sətirlərin (dataların) neçə faizinin təsadüfi seçiləcəyi.
    # 0.8 (yəni 80%) etmək həm sürəti artırır, həm də variance-ı (overfitting-i) azaldır.
    subsample=0.8,

    # colsample_bytree (Default: 1.0)
    # Hər ağac qurularkən sütunların (feature) neçə faizinin təsadüfi seçiləcəyi.
    # Bu da modelin fərqli aspektləri öyrənməsini təmin edir. 0.7 - 0.9 arası optimaldır.
    colsample_bytree=0.8,

    # =================================================================
    # 4. REQULYARİZASİYA (REGULARIZATION)
    # =================================================================

    # reg_alpha / alpha (Default: 0)
    # L1 Requlyarizasiyası (Lasso). Önəmsiz sütunların çəkisini tam sıfıra endirməyə çalışır.
    # Datan çox yüksək ölçülüdürsə (çox sütun varsa) işə yarayır.
    reg_alpha=0.01,

    # reg_lambda / lambda (Default: 1)
    # L2 Requlyarizasiyası (Ridge). Çəkiləri cəzalandıraraq kiçildir, modeli daha hamar edir.
    reg_lambda=1.0,

    # =================================================================
    # 5. İŞLƏMƏ SÜRƏTİ VƏ BALANS (PERFORMANCE & BALANCE)
    # =================================================================

    # scale_pos_weight (Default: 1)
    # Balansı kəskin pozulmuş datalar (Imbalanced data) üçün "Həyat Qurtaran" parametrdir.
    # Məsələn: 100 xəstə (Positive), 900 sağlam (Negative) varsa, 
    # scale_pos_weight = 900 / 100 = 9 yazmalısan ki, model azlıqda olan xəstələri ciddiyə alsın.
    scale_pos_weight=1,

    # tree_method (Default: 'auto')
    # Ağac qurma alqoritmi. Əgər datanda 10.000-dən çox sətir varsa, bunu mütləq 'hist' et. 
    # 'hist' (histogram) qat-qat daha sürətlidir və eyni keyfiyyəti verir.
    tree_method='hist',

    # n_jobs (Default: None)
    # Kompüterinin/Serverinin neçə nüvəsindən (core) istifadə ediləcəyi. 
    # -1 etsən bütün gücünü XGBoost-a verəcək.
    n_jobs=-1,

    # random_state
    # Kodu hər işlədəndə eyni rəqəmləri almaq üçün sabit bir toxum (seed).
    random_state=42
)
```

```python
xgb = XGBClassifier(n_estimators = 100)

xgb.fit(X_train,y_train)
```

```python
y_pred = xgb.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix,accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```
