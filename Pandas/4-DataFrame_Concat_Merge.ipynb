```python
#merge
```

```python
df_merge1 =  pd.read_csv('7-merge_data1.csv')
```

```python
df_merge1
```

```python
df_merge2 =  pd.read_csv('7-merge_data2.csv')
```

```python
df_merge2
```

```python
#merge-outer join
```

```python
df_merged_outer = pd.merge(df_merge1,df_merge2,on = "Employee_ID",how = "outer")#1  on - nəyə görə birləşdirməsini müəyyən edir
# how - necƏ birləşdirəcəyini deyir outer olanda Hər hansı sütun və ya  sətirdə null olanda həmin sətir və sütunuda gətirir,
#Hər hansı idlər kəsişməyəndə onlarıda gətirir kəsişməyəndə ikinci mergdə olana NAN yazır əgər hüceyrə boşdursa nan yazır
```

```python
df_merged_outer
```

```python
#merge - left join
```

```python
df_merged_left = pd.merge(df_merge1,df_merge2,on = "Employee_ID",how = "left") #left isə birinci cədvəldəkilərin hamısın al 
#ikinci cədvəldən isə yalnız birinci ilə kəsişənləri al
```

```python
df_merged_left 
```

```python
df_merged_right = pd.merge(df_merge1,df_merge2,on = "Employee_ID",how = "right")
```

```python
df_merged_right #Bu səfərdə ikinci cədvəli əsas götürür yəni ikinci cədvəldəki bütün məlumatları gəririr 1 cidən isə yalnız ikinci ilə kəsişənləri gətirir
```

```python
df_merged_inner = pd.merge(df_merge1,df_merge2,on = "Employee_ID",how = "inner") #Hər ikisində ortaq olan dataları gətirir
```

```python
df_merged_inner
```
