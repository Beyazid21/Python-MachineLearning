```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("26-customer_data.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
sns.scatterplot(data= df,x='Annual_Income',y = 'Spending_Score')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test = train_test_split(df,test_size=0.2,random_state=15)
```

```python
# we can use normalization rather than standardscaler in algorithms such as knn, kmeans
# neural networks, or when we want to normalize data and put the datas between 0-1
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python

from sklearn.cluster import KMeans
```

```python
## Elbow method To select K Value
wcss=[]
for k in range(1,11): #try k numbers from 1 to 11
    kmeans=KMeans(n_clusters=k,init="k-means++")  #default value of init is k-means++ the other value is random
    kmeans.fit(X_train_scaled)
    wcss.append(kmeans.inertia_)
```

```python
kmeans.labels_
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
plt.xticks(range(1,11))
plt.xlabel("Number of Clustrers")
plt.ylabel("WCSS")
plt.show()
```

```python
kmeans=KMeans(n_clusters=3,init="k-means++")
```

```python
kmeans.fit(X_train_scaled)
```

```python
y_pred=kmeans.predict(X_test_scaled)
```

```python
sns.scatterplot(data=pd.DataFrame(X_test_scaled, columns=X_test.columns), x="Annual_Income", y="Spending_Score", hue=y_pred)
plt.show()
```

```python
# Other alternative
# silhoutte score
from sklearn.metrics import silhouette_score
```

```python

silhouette_coefficients=[]
for k in range(2,11): # it needs at least 2 cluster labels so we are not trying 1
    kmeans=KMeans(n_clusters=k,init="k-means++")
    kmeans.fit(X_train_scaled)
    score=silhouette_score(X_train_scaled,kmeans.labels_) 
    silhouette_coefficients.append(score)
```

```python
silhouette_coefficients
```

```python
plt.plot(range(2,11),silhouette_coefficients)
plt.xticks(range(2,11))
plt.xlabel("Number of Cluters")
plt.ylabel("Silhoutte Coeffecient")
plt.show()
```
